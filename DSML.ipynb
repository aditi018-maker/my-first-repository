```python
df["Recovered"]=df["Cured"]
```

```python
df['Active'] = df['Confirmed'] - df['Cured'] - df['Deaths']
df.head()
```

```python
static_data = (
    df.groupby("State/UnionTerritory")
    [["Confirmed", "Deaths", "Cured","Active"]]
    .max()
    .reset_index()
)
```

```python
static_data["Recovery_Rate"] = static_data["Cured"] / static_data["Confirmed"]
```

```python
static_data["Death_Rate"] = static_data["Deaths"] / static_data["Confirmed"]
```

```python
import pandas as pd

top10 = (
    static_data
    .sort_values(by="Confirmed", ascending=False)
    .head(10)
)

print(top10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("darkgrid")

sns.barplot(
    data=top10,
    x="State/UnionTerritory",
    y="Confirmed"
)
```

```python
daily_cases =(
    df.groupby("Date")
    ["Confirmed"]
    .max()
    .reset_index()
)
sns.lineplot(
    data=daily_cases,
    x="Date",
    y="Confirmed",
    color = "red"
)
plt.title("covid-19confirmed cases over time")
plt.xlabel("Date")
plt.ylabel("Confirmed Cases")
plt.xticks(rotation=45)
plt.show()
```
