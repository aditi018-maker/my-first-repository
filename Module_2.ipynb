```python
import pandas as pd
```

```python
data = {
    "Employees": ["E01","E02","E03","E04","E05","E06","E07","E08","E09","E10"],
    "Salary": [28000,30000,32000,35000,36000,40000,42000,45000,150000,250000]
}
df = pd.DataFrame(data)
df
```

```python
df.info()
```

```python
mean_salary = df["Salary"].mean()
print("Mean Salary:", mean_salary)
```

```python
median_salary = df["Salary"].median()
print("Median Salary:", median_salary)
```

```python
Mode_Salary = df["Salary"].mode()
print("Mode Salary:", Mode_Salary)
```

```python
# Range maximum - minimum

salary_range = df["Salary"].max() - df["Salary"].min()
print("Salary Range:", salary_range)
```

```python
# variance
variance = df["Salary"].var()
print("Variance :",variance)
```

```python
# standard deviation

std_dev = df["Salary"].std()
print("Standard Deviation :",std_dev)
```

```python
# percentlies
print("50th percentile",
      df["Salary"].quantile(0.5))
```

```python
#quantile
q1 = df["Salary"].quantile(0.25)
q2 = df["Salary"].quantile(0.50)
q3 = df["Salary"].quantile(0.75)

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
```

```python
iqr = q3 - q1
print("IQR :",iqr)
```

```python
lower_bound = q1 - 1.5 * (q3 - q1)
upper_bound = q3 + 1.5 * (q3 - q1)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

# find the outliers

outliers = df[(df["Salary"] < lower_bound) | (df["Salary"] > upper_bound)]
print("Outliers:")
print(outliers)
```

```python
import numpy as np
from scipy.stats import ttest_rel

# Ensure before and after are defined in case they were not run
before = np.array([9800, 10200, 10100, 9900, 10500, 10000, 9700, 10300, 10100, 9900, 10200, 9800, 10400, 10000])
after = np.array([10800, 11000, 10700, 10900, 11100, 10600, 11200, 10800, 10900, 9900, 10700, 11100, 10900, 10800])

# Calculate the p-value
test_statistic, p_value = ttest_rel(after, before, alternative="greater")

alpha = 0.05

if p_value < alpha:
    print("Reject the H0")
    print("Result is statistically significant")
else:
    print("Fail to reject H0")
    print("Result is not statistically significant")
# p-value < 0.05 --> reject
```

```python
import numpy as np
from scipy.stats import ttest_rel

before_delivery = np.array([31, 28, 30, 32, 26])

after_delivery = np.array([27, 28, 29, 26, 30])

stat, p = ttest_rel(
    before_delivery,
    after_delivery,
    alternative="greater"
)

print("Test statistic:", stat)
print("p-value:", p)
```

```python
if p < 0.05:

  print("Reject H0")
  print("Evidence supports a decrease in delivery time")
else:
  print("Fail to reject H0")
  print("Insufficient evidence of a decrease in delivery time")
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

# Rows = course preference
# column = learning format
observed = pd.DataFrame(
    {
        "Online": [40,25],
        "Classroom":[20,35]
    },
    index=["Data Science" , "Web Development"]


)
print("Obererved Student counts")
print(observed)
```

```python
chi2, p_value, dof, expected = chi2_contingency(observed)
print("\nExpected counts if the variable were independent")
print(pd.DataFrame(
    expected,
    index = observed.index,
    columns = observed.columns,
).head(2))
print("\nchi-squared test statistic", round(chi2, 3))
print("Degree of freedom", dof)
print("p-value", round(p_value, 4))

if p_value < 0.05:
  print("Decision: Evidence of an association between course preference and learning format")
else:
  print("Insufficient Evidence")
```
